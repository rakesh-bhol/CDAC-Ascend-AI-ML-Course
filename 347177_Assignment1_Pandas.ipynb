```python
import pandas as pd
```

```python
# Q1: Read the file it_services_tickets_raw.csv into a DataFrame called df.
df = pd.read_csv('it_services_tickets_raw.csv')
# Explanation: Reads the CSV dataset from disk and stores it as a pandas DataFrame.
```

```python
# Q2: Show the first 10 rows.
print(df.head(10))
# Explanation: Displays the first 10 rows.
```

```python
# Q3: Show the last 10 rows.
print(df.tail(10))
# Explanation: Displays the final 10 rows at the bottom of the dataset.
```

```python
# Q4: Print how many rows and how many columns the table has.
print(f"Dimensions: {df.shape[0]} rows and {df.shape[1]} columns")
# Explanation: Displays the dimensions (rows, columns) of the DataFrame.
```

```python
# Q5: Print the names of all the columns.
print(df.columns.tolist())
# Explanation: Displays the exact raw column headers.
```

```python
# Q6: Print the data type of each column.
print(df.dtypes)
# Explanation: Displays pandas data types for each column.
```

```python
# Q7: Run df.info() and write one line about what it tells you.
df.info()
# Explanation: Summary showing column data types, total non-null counts, and memory consumption.
```

```python
# Q8: Run df.describe() and look at the numeric columns.
print(df.describe())
# Explanation: Displays statistical summaries (mean, min, max, quartiles) for all numeric columns.
```

```python
# Q9: Change all column names to lowercase, and replace spaces with underscores.
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
print(df.columns.tolist())
# Explanation: Standardizes column header names.
```

```python
# Q10: Print how many missing values each column has.
print(df.isna().sum())
# Explanation: Calculates the count of null values per column.
```

```python
# Q11: Print the total number of missing values in the whole table.
print("The total number of missing values in the whole table:", df.isna().sum().sum())
# Explanation: Aggregates total missing entries across the entire dataset.
```

```python
# Q12: Replace the text values NA, N/A, null and - with proper missing values.
missing_representations = ["NA", "N/A", "null", "NULL", "-", "?", "missing", "nan"]
df = df.replace(missing_representations, None)
print("Standardized missing values count:", df.isna().sum().sum())
# Explanation: Replaces string placeholders of missing data with None.
```

```python
# Q13: Count how many duplicate rows there are.
print("Duplicate rows count:", df.duplicated().sum())
# Explanation: Identifies the exact count of fully identical records across rows.
```

```python
# Q14: Remove the duplicate rows and print the new number of rows.
df = df.drop_duplicates()
print("New row count after removing duplicates:", len(df))
# Explanation: Drops all duplicate entries and retains unique records.
```

```python
# Q15: Show only three columns: ticket id, client name and status.
print(df[['ticket_id', 'client_name', 'status']])
# Explanation: Filters and show only the three selected columns.
```

```python
# Q16: Show rows 5 to 15 using iloc.
display(df.iloc[5:16])
# Explanation: Displays rows from index 5 up to 15 (inclusive).
```

```python
# Q17: Show the row at index 100 using loc.
display(df.loc[100])
# Explanation: Fetches the record corresponding explicitly to label index 100.
```

```python
# Q18: Show all the tickets where priority is P1.
display(df[df['priority'] == 'P1'])
# Explanation: Filters the dataset for priority P1.
```

```python
# Q19: Show all the tickets where status is Open.
display(df[df['status'].astype(str).str.lower() == 'open'])
# Explanation: Returns all records currently in an open status state.
```

```python
# Q20: Count how many tickets there are for each status.
print(df['status'].value_counts(dropna=False))
# Explanation: Counts total ticket occurrences grouped by status.
```

```python
# Q21: Count how many tickets there are for each priority.
print(df['priority'].value_counts(dropna=False))
# Explanation: Frequency distribution of tickets across priority levels.
```

```python
# Q22: Count how many tickets there are for each service line.
print(df['serviceline'].value_counts(dropna=False))
# Explanation: Displays ticket volumes across different organizational service lines.
```

```python
# Q23: Find how many different clients there are.
print("Unique clients count:", df['client_name'].nunique())
# Explanation: Calculates the total number of unique client accounts.
```

```python
# Q24: Print the list of all countries in the data.
print(df['client_country'].dropna().unique().tolist())
# Explanation: Lists all unique client operating countries represented in the dataset.
```

```python
 # Q25: Convert the hours logged column to a number.
df['hours_logged'] = pd.to_numeric(df['hours_logged'], errors='coerce')
print("hours_logged dtype:", df['hours_logged'].dtype)
# Explanation: Casts the hours_logged column into numeric floating/integer format.
```

```python
# Q26: Convert the billing rate column to a number. Remove the $ sign first.
df['billing_rate_(usd)'] = df['billing_rate_(usd)'].astype(str).str.replace('$', '', regex=False).str.strip()
df['billing_rate_(usd)'] = pd.to_numeric(df['billing_rate_(usd)'], errors='coerce')
print("billing_rate_(usd) dtype:", df['billing_rate_(usd)'].dtype)
# Explanation: Strips currency symbols before converting rates into numeric format.
```

```python
# Q27: Convert the CSAT score column to a number.
df['csat_score'] = pd.to_numeric(df['csat_score'], errors='coerce')
print("csat_score dtype:", df['csat_score'].dtype)
# Explanation: Ensures customer satisfaction scores are represented as float numbers.
```

```python
# Q28: Print the average, the smallest and the largest value of hours logged.
print(f"Mean: {df['hours_logged'].mean():.2f}, Min: {df['hours_logged'].min()}, Max: {df['hours_logged'].max()}")
# Explanation: Computes fundamental summary statistics for hours logged across tickets.
```

```python
# Q29: Sort the table by hours logged from highest to lowest and show the top 10 rows.
display(df.sort_values(by='hours_logged', ascending=False).head(10))
# Explanation: Orders the DataFrame in descending order of effort logged.
```

```python
# Q30: Show the 5 tickets with the highest billable amount.
df['billable_amount_usd'] = pd.to_numeric(df['billable_amount_usd'], errors='coerce')
display(df.nlargest(5, 'billable_amount_usd'))
# Explanation: Retrieves the top 5 tickets based on revenue generated.
```

```python
# Q31: Remove the extra spaces from the client name column.
df['client_name'] = df['client_name'].astype(str).str.strip()
print("Client names stripped of leading/trailing whitespace.")
# Explanation: Trims leading and trailing whitespace from client string names.
```

```python
# Q32: Change the client name column to title case.
df['client_name'] = df['client_name'].str.title()
print(df['client_name'].head())
# Explanation: Capitalizes the first letter of each word in client names.
```

```python
# Q33: Convert the ticket created column to a date.
df['ticket_created_at'] = pd.to_datetime(df['ticket_created_at'], format='mixed', errors='coerce')
print("ticket_created_at dtype:", df['ticket_created_at'].dtype)
# Explanation: Converts mixed string timestamp formats into standardized Pandas Datetime format.
```

```python
# Q34: Add a new column that holds the year of the ticket.
df['ticket_year'] = df['ticket_created_at'].dt.year
print(df[['ticket_created_at', 'ticket_year']].head())
# Explanation: Extracts the calendar year from the ticket creation timestamp into a new column.
```

```python
# Q35: Add a new column that holds the month of the ticket.
df['ticket_month'] = df['ticket_created_at'].dt.month
print(df[['ticket_created_at', 'ticket_month']].head())
# Explanation: Extracts the numerical calendar month (1-12) from the creation date.
```

```python
# Q36: Count how many tickets were created in each year.
print(df['ticket_year'].value_counts().sort_index())
# Explanation: Computes total volume of tickets submitted per year.
```

```python
# Q37: Add a new column called total equal to billing rate multiplied by hours logged.
df['total'] = df['billing_rate_(usd)'] * df['hours_logged']
display(df[['billing_rate_(usd)', 'hours_logged', 'total']].head())
# Explanation: Creates a calculated total amount based on rate and effort.
```

```python
# Q38: Fill the missing values in the CSAT score column with the average CSAT score.
avg_csat = df['csat_score'].mean()
df['csat_score'] = df['csat_score'].fillna(avg_csat)
print(f"Missing CSAT values filled with mean: {avg_csat:.2f}")
# Explanation: Imputes missing satisfaction scores using the dataset-wide arithmetic mean.
```

```python
# Q39: Drop the rows where the client name is missing.
df = df.dropna(subset=['client_name'])
print("New row count after dropping missing client names:", len(df))
# Explanation: Cleans dataset by removing tickets unassigned to any valid client.
```

```python
# Q40: Group by service line and find the average hours logged.
display(df.groupby('serviceline')['hours_logged'].mean())
# Explanation: Displays the mean work effort required per service line department.
```

```python
# Q41: Group by priority and count the tickets.
display(df.groupby('priority').size())
# Explanation: Shows aggregate ticket counts split across distinct priority tiers.
```

```python
# Q42: Group by client name and find the total billable amount.
display(df.groupby('client_name')['billable_amount_usd'].sum())
# Explanation: Calculates total cumulative revenue generated per client account.
```

```python
# Q43: Find the top 5 clients by number of tickets.
print(df['client_name'].value_counts().head(5))
# Explanation: Lists five client entities submitting the highest volume of tickets.
```

```python
# Q44: Find the average CSAT score for each priority.
display(df.groupby('priority')['csat_score'].mean())
# Explanation: Evaluates customer satisfaction scores across ticket priority levels.
```

```python
# Q47: Show only the tickets where hours logged is more than 50.
display(df[df['hours_logged'] > 50])
# Explanation: Filters dataset to highlight tickets taking over 50 hours of work.
```

```python
# Q48: Show the tickets where status is Closed and priority is P1.
display(df[(df['status'].astype(str).str.lower() == 'closed') & (df['priority'] == 'P1')])
# Explanation: Extracts completed high-priority (P1) tickets.
```

```python
# Q49: In the SLA breached column, change Yes, Y, TRUE and 1 to True, and everything else to False.
truthy = ['yes', 'y', 'true', '1', 1, True]
df['sla_breached'] = df['sla_breached'].astype(str).str.strip().str.lower().isin(truthy)
print(df['sla_breached'].value_counts())
# Explanation: Normalizes boolean breach flag into strict True/False values.
```

```python
# Q50: Save your cleaned table to a new CSV file.
df.to_csv('it_services_tickets_cleaned.csv', index=False)
print("Cleaned DataFrame successfully saved to 'it_services_tickets_cleaned.csv'.")
# Explanation: Exports clean data into a CSV file without writing pandas index column.
```
